```python
import pandas as pd
```

```python
train_file_path = "train-data.tsv"
test_file_path = "valid-data (1).tsv"


train_data = pd.read_csv(
    train_file_path,
    sep='\t',
    header=None,
    names=['label', 'message']
)

test_data = pd.read_csv(
    test_file_path,
    sep='\t',
    header=None,
    names=['label', 'message']
)

print("Training data:")
print(train_data.head())

print("\nTesting data:")
print(test_data.head())

print("\nTraining shape:", train_data.shape)
print("Testing shape:", test_data.shape)
```

```python
train_data['label'] = train_data['label'].map({
    'ham': 0,
    'spam': 1,

})

test_data['label'] = train_data['label'].map({
    'ham': 0,
    'spam': 1,

})

print(train_data.head())
print('\n Label counts:')
print(train_data['label'].value_counts())
```

```python
train_messages = train_data['message'].values
train_labels = train_data['label'].values

test_messages = test_data['message'].values
test_labels = test_data['label'].values

print("Training messages:", len(train_messages))
print("Training labels:", len(train_labels))
print("Testing messages:", len(test_messages))
print("Testing labels:", len(test_labels))
```

```python
from tensorflow.keras import layers # type: ignore
print('layers imported successfully')
```

```python
max_features = 10000
sequence_length = 100

vectorize_layer = layers.TextVectorization(
    max_tokens=max_features,
    output_mode='int',
    output_sequence_length=sequence_length
)

vectorize_layer.adapt(train_messages)

print("Text vectorizer ready!")
```

```python
from tensorflow import keras # type: ignore
print('Improt keras successfully')
```

```python
model = keras.Sequential([
    vectorize_layer,
    layers.Embedding(max_features, 16),
    layers.GlobalAveragePooling1D(),
    layers.Dense(16, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

print("Model created successfully!")
```

```python
history = model.fit(
    train_messages,
    train_labels,
    validation_split=0.2,
    epochs=10,
    batch_size=32,
    class_weight = {0: 1.0, 1: 3.0}
)
```

```python
test_data['label'] = test_data['label'].map({
    'ham': 0,
    'spam': 1
})

test_messages = test_data['message'].values
test_labels = test_data['label'].values

print(test_labels[:20])
print("Test messages:", len(test_messages))
print("Test labels:", len(test_labels))
```

```python
test_loss, test_accuracy = model.evaluate(
    test_messages,
    test_labels,
    verbose=0
)

print("Test Accuracy:", test_accuracy)
```

```python
def predict_message(message):
    prediction = model.predict(
        tf.constant([message]),
        verbose=0
    )[0][0]

    if prediction >= 0.5:
        return [float(prediction), "spam"]
    else:
        return [float(1 - prediction), "ham"]
```

```python
print(predict_message("Congratulations! You have won a free prize. Call now!"))
print(predict_message("Hey, are you coming to college tomorrow?"))
print(predict_message("URGENT! You have won 1000 pounds. Claim now!"))
```
